Find a difference between python list and numpy

- **Type & homogeneity**: Python lists can store mixed data types; NumPy arrays are homogeneous and store elements of a single data type.  
- **Performance**: NumPy arrays are optimized in C for vectorized operations and are typically faster and more memory-efficient than Python lists for numerical work.  
- **Operations**: NumPy arrays support element-wise arithmetic, broadcasting, and rich mathematical functions out of the box; Python lists require explicit loops or comprehensions for similar behavior.  
- **Shape**: NumPy arrays natively handle multi-dimensional data; Python lists need nested structures to emulate higher dimensions.  
- **Memory layout**: NumPy arrays are contiguous in memory (when possible), enabling faster computations; Python lists are arrays of pointers, leading to additional overhead.
